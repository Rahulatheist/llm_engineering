```python
# imports
from openai import OpenAI
import os
import json
from dotenv import  load_dotenv
from IPython.display import display, update_display,Markdown
```

```python
# constants

OPENROUTER_API_URL = "https://openrouter.ai/api/v1"
ollama = OpenAI(base_url=OPENROUTER_API_URL, api_key=os.getenv("OPENROUTER_API_KEY"))
MODEL = "nvidia/nemotron-3-ultra-550b-a55b:free"
```

```python
# set up environment
system_prompt = """
You are a Senior software engineer with 20 years of experience. 
You are an expert in Python, JavaScript, and cloud computing. 
You are also an expert in LLMs and prompt engineering. 
You have a deep understanding of the latest advancements in AI and machine learning. 
You are a great teacher and can explain complex concepts in a simple way.

"""
```

```python
def get_user_prompt(user_input):
    return f"""
    Explain the following user input in detail, providing insights, examples, and any relevant information that would help a reader understand the context and implications of the input.
    User input: {user_input}
    
    """
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
```

```python
def get_message_stream(user_input):
    return ollama.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": get_user_prompt(user_input)}
        ]
    )
```

```python
# Get gpt-4o-mini to answer, with streaming
def get_detailed_answer(user_input):
    stream = get_message_stream(user_input)
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
            response += chunk.choices[0].delta.content or ''
            update_display(Markdown(response), display_id=display_handle.display_id)
```

```python
# Get Llama 3.2 to answer
get_detailed_answer(question)
```
