NF4 (4-bit NormalFloat) quantization is an information-theoretically optimal 4-bit data type that spaces its 16 quantization levels to match a standard normal (Gaussian) distribution rather than using uniform spacing. [1, 2]  
How NF4 Works 

• Non-Uniform Spacing: Because neural network and transformer weights naturally cluster tightly around zero in a bell-curve shape, NF4 places its 16 discrete levels at the quantiles of a standard normal distribution. 
• Equal Probability Mass: Each adjacent pair of levels bounds a region containing an equal fraction of the total probability mass. More precision is allocated near zero, while the sparse tails get wider bins. 
• Quantile Quantization: Input weights are scaled by the block absolute maximum (), mapped to the nearest quantile in the lookup table, and stored compactly as 4-bit values. Dequantization happens on the fly during runtime. [3, 4]  

Why NF4 Matters for QLoRA 

• Near-Baseline Accuracy: By matching the true distribution of pretrained weights, NF4 recovers significant perplexity (accuracy) compared to standard uniform INT4 quantization at the same 4-bit limit. 
• Efficient Fine-Tuning: Introduced via the  QLoRA Paper  (implemented heavily via Hugging Face bitsandbytes), NF4 enables stable 4-bit base model training and parameter fine-tuning on a single consumer GPU. 
• Double Quantization: To prevent per-block scaling factors from adding heavy memory overhead (which can waste roughly 0.5 bits per weight), double quantization re-quantizes those scaling constants into an 8-bit format, saving an extra ~0.4 bits per parameter with zero drop in performance.
